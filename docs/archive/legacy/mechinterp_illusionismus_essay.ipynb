## What was changed mathematically?

Addition: $r′=r+\alpha m v$, where $v$ is unit length and $m$ is a development-only reference norm.

Feature ablation: $r′=r-f_i(r)d_i$, preserving the original SAE reconstruction remainder. Negative direction addition is a different intervention.

The inferential unit is an independent scenario, not each paraphrase or seed. See `docs/PROTOCOL.md` for assumptions and limitations.
